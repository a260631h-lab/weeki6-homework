**この組織が知りたいこと：**  
高齢者支援のための新規拠点を設置・人員配置するにあたり、高齢者の「絶対的な人数が多い地域」と「全体に対する割合（高齢化率）が高い地域」にはどのような違いがあるか。

---

```python
import csv

with open("week6/japan.csv", encoding="utf-8-sig", newline="") as file:
    data = list(csv.DictReader(file))

print(len(data))
```

## データの型変換と指標（高齢者割合）の計算

文字列として読み込まれた「総人口」と「65歳以上人口」を整数型（int）に変換し、各都道府県の**高齢者割合（%）**を新しく計算してデータに追加します。

```python
# 数値への変換と割合の計算
for row in data:
    population = int(row["総人口"])
    seniors = int(row["65歳以上人口"])
    
    # 割合の計算 (小数第2位で四捨五入)
    senior_share = round((seniors / population) * 100, 1)
    
    # 辞書に計算結果と数値データを保存
    row["総人口_num"] = population
    row["65歳以上_num"] = seniors
    row["高齢者割合"] = senior_share

# 動作確認
sample = data[0]
print(f"{sample['都道府県']}: 高齢者数 {sample['65歳以上_num']:,}人 / 割合 {sample['高齢者割合']}%")
```

## 分析1：高齢者「人数」のトップ5とワースト5

まずは支援を必要とする「対象者の絶対数」が多い都道府県を把握するため、65歳以上人口の人数順で並べ替えます。

```python
# 高齢者「人数」で降順（大きい順）にソート
sorted_by_count = sorted(data, key=lambda x: x["65歳以上_num"], reverse=True)

print("--- 高齢者「人数」が多い都道府県 TOP 5 ---")
for i, row in enumerate(sorted_by_count[:5], 1):
    print(f"{i}位: {row['都道府県']} - {row['65歳以上_num']:,} 人 (割合: {row['高齢者割合']}%)")

print("\n--- 高齢者「人数」が少ない都道府県 TOP 5 ---")
for i, row in enumerate(sorted_by_count[-5:], 1):
    print(f"{i}位: {row['都道府県']} - {row['65歳以上_num']:,} 人 (割合: {row['高齢者割合']}%)")
```

**【人数の傾向】**  
人数上位は東京都・神奈川県・大阪府などの大都市圏が占めています。人口自体が多いため、高齢者数も絶対数として膨大になります。

## 分析2：高齢者「割合（高齢化率）」のトップ5とワースト5

次に、地域コミュニティの高齢化の深刻度を測るため、「総人口に対する65歳以上人口の割合」で並べ替えます。

```python
# 高齢者「割合」で降順（大きい順）にソート
sorted_by_share = sorted(data, key=lambda x: x["高齢者割合"], reverse=True)

print("--- 高齢者「割合」が高い都道府県 TOP 5 ---")
for i, row in enumerate(sorted_by_share[:5], 1):
    print(f"{i}位: {row['都道府県']} - {row['高齢者割合']}% (人数: {row['65歳以上_num']:,}人)")

print("\n--- 高齢者「割合」が低い都道府県 TOP 5 ---")
for i, row in enumerate(sorted_by_share[-5:], 1):
    print(f"{i}位: {row['都道府県']} - {row['高齢者割合']}% (人数: {row['65歳以上_num']:,}人)")
```

## 結論と組織への提言

### 問いに対する答え
「高齢者の人数が多い地域」と「割合が高い地域」は**全く異なる自治体群**となりました。
- **人数上位**：大都市圏（東京・神奈川・大阪など）
- **割合上位**：地方都市・過疎地域（秋田・高知など）

### 組織への具体的な施策提言
1. **大都市圏（大規模ケア拠点モデル）**
   - 絶対数が多いため、介護サービスや施設などの「物理的なキャパシティ」の拡充が最優先。
2. **地方・過疎地域（移動・見守り支援モデル）**
   - 割合が高いため地域全体の担い手が不足している。見守りネットワークや訪問型支援など、コミュニティ維持施策にリソースを集中させるべき。
